### Task 1 — Create Your Own DataFrame
Create a DataFrame called `students` with **at least 5 students** and these columns:
- `Name`
- `Age`
- `Subject` (their favourite subject)
- `Marks`

Use any names/values you like.

```python
students = {
    "Name": ["Bilal", "Ahmed", "Ali", "Hassan"],
    "Age": [24, 26, 29, 27],
    "Subject": ["Data structures", "Networks", "Web dev", "Statistics"],
    "Marks": [82, 89, 65, 74]
}
df = pd.DataFrame(students)
df
```

### Task 4 — Filter Rows
Display only the students who scored **more than 70** marks.

```python
high_scores = df[df["Marks"] > 70]
high_scores
```

### Task 5 — Add a Column
Add a new column called `Grade`. Set it to `"A"` if Marks >= 80, otherwise `"B"`.

Hint: you can first set the whole column to `"B"`, then use `.loc` to change specific rows to `"A"` — just like we did with Bilal's city earlier.

```python
df["Grade"] = "B"
df
```

```python
df.loc[df["Marks"] >= 80, "Grade"] = "A"
df
```

### Task 6 — Sort
Sort your `students` DataFrame by `Marks`, from **highest to lowest**.

```python
sort = df.sort_values("Marks", ascending=False)
sort
```

### Task 7 — Handle Missing Data
1. Change one student's `Age` value to `np.nan` (missing) using `.loc`.
2. Check how many missing values are in the `Age` column.
3. Fill the missing value with the average age.

```python
df.loc[df["Name"] == "Hassan", "Age"] = np.nan
df
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

### Task 8 — Group and Summarize
Use `groupby` to find the **average Marks** for each `Subject`.

```python
df.groupby("Subject")["Marks"].mean()
```
